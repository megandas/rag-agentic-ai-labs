```python
import gradio as gr
from transformers import BlipProcessor, BlipForConditionalGeneration

processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

def generate_caption(image):
    # Now directly using the PIL Image object
    inputs = processor(images=image, return_tensors="pt")
    outputs = model.generate(**inputs)
    caption = processor.decode(outputs[0], skip_special_tokens=True)
    return caption

def caption_image(image):
    """
    Takes a PIL Image input and returns a caption.
    """
    try:
        caption = generate_caption(image)
        return caption
    except Exception as e:
        return f"An error occurred: {str(e)}"

iface = gr.Interface(
    fn=caption_image,
    inputs=gr.Image(type="pil"),
    outputs="text",
    title="Image Captioning with BLIP",
    description="Upload an image to generate a caption."
)

iface.launch(server_name="127.0.0.1")
```

```python
import torch
import requests
from torchvision import transforms

# Download human-readable labels for ImageNet
response = requests.get("https://git.io/JJkYN")
labels = [l.strip() for l in response.text.split("\n") if l.strip()]

# Define image preprocessing (IMPORTANT for ResNet)
transform = transforms.Compose([ transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

def predict(inp):
    # preprocess image
    inp = transform(inp).unsqueeze(0)

    # ensure model runs in inference mode
    with torch.no_grad():
        prediction = torch.nn.functional.softmax(model(inp)[0], dim=0)

    # map predictions to labels
    confidences = {
        labels[i]: float(prediction[i])
        for i in range(len(labels))
    }

    return confidences
```

```python
import gradio as gr

gr.Interface(fn=predict,
       inputs=gr.Image(type="pil"),
       outputs=gr.Label(num_top_classes=3),
       examples=["/content/lion.jpg", "/content/cheetah.jpg"]).launch()
```

```python
from transformers import pipeline

# Load Qwen locally
chatbot = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=256,
    temperature=0.5,
    do_sample=True,
    return_full_text=False,
    clean_up_tokenization_spaces=False
)

print("✅ Qwen chatbot loaded!")
```

```python
def generate_response(prompt_txt):
    response = chatbot(prompt_txt)[0]["generated_text"]
    return response
```

```python
import gradio as gr

chat_application = gr.Interface(
    fn=generate_response,
    inputs=gr.Textbox(
        label="Input",
        lines=2,
        placeholder="Type your question here..."
    ),
    outputs=gr.Textbox(label="Output"),
    title="Qwen AI Chatbot",
    description="Ask any question and the chatbot will answer using the local Qwen model."
)
```

```python
chat_application.launch(
    share=True
)
```

```python
import warnings
warnings.filterwarnings("ignore")

import gradio as gr

from transformers import pipeline
from sentence_transformers import SentenceTransformer

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_classic.chains import RetrievalQA
from langchain_community.llms import HuggingFacePipeline
```

```python
llm_pipeline = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=256,
    temperature=0.5,
    do_sample=False,
    return_full_text=False,
    clean_up_tokenization_spaces=False
)

llm = HuggingFacePipeline(pipeline=llm_pipeline)

print("✅ Local Qwen model loaded!")
```

```python
def document_loader(file_path):
    loader = PyPDFLoader(file_path)
    documents = loader.load()
    return documents
```

```python
def text_splitter(documents):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=100,
        length_function=len
    )

    chunks = splitter.split_documents(documents)
    return chunks
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("✅ Embedding model loaded!")
```

```python
def create_vector_db(chunks):
    vectordb = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings
    )
    return vectordb
```

```python
def create_retriever(file_path):
    docs = document_loader(file_path)
    chunks = text_splitter(docs)
    vectordb = create_vector_db(chunks)

    return vectordb.as_retriever(
        search_kwargs={"k":3}
    )
```

```python
def retriever_qa(file_path, query):

    retriever = create_retriever(file_path)

    qa = RetrievalQA.from_chain_type(
        llm=llm,
        chain_type="stuff",
        retriever=retriever,
        return_source_documents=False
    )

    result = qa.invoke({"query": query})

    return result["result"]
```

```python
rag_application = gr.Interface(
    fn=retriever_qa,

    inputs=[
        gr.File(
            label="Upload PDF",
            file_types=[".pdf"],
            type="filepath"
        ),

        gr.Textbox(
            label="Ask a Question",
            lines=2,
            placeholder="Ask anything from the uploaded PDF..."
        )
    ],

    outputs=gr.Textbox(label="Answer"),

    title="📄 PDF RAG Chatbot (Local)",

    description="""
    Upload any PDF and ask questions about it.
    Uses LangChain + Chroma + Qwen locally.
    """
)
```

```python
rag_application.launch(share=True)
```
