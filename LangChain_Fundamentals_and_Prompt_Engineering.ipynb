```python
llm = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=128,
    temperature=0.3,
    clean_up_tokenization_spaces=False,
)
```

```python
def ask_llm(prompt):
    messages = [{"role": "user", "content": prompt}]
    response = llm(messages)
    return response[0]["generated_text"][-1]["content"]
```

```python
from transformers import pipeline

# Model 1 (Granite replacement)
qwen = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=128,
    temperature=0.2,
    clean_up_tokenization_spaces=False,
)

# Model 2 (Llama replacement)
smollm = pipeline(
    "text-generation",
    model="HuggingFaceTB/SmolLM2-360M-Instruct",
    max_new_tokens=128,
    temperature=0.8,
    clean_up_tokenization_spaces=False,
)
```

```python
def generate(pipe, prompt):
    messages = [{"role": "user", "content": prompt}]
    output = pipe(messages)
    return output[0]["generated_text"][-1]["content"]
```

```python
def chunk_statistics(chunks):
    lengths = [len(chunk.page_content) for chunk in chunks]

    print(f"Total chunks      : {len(chunks)}")
    print(f"Average chunk size: {sum(lengths)/len(lengths):.1f}")
    print(f"Minimum chunk size: {min(lengths)}")
    print(f"Maximum chunk size: {max(lengths)}")
```

```python
print("CharacterTextSplitter Statistics")
chunk_statistics(character_chunks)

print("\nRecursiveCharacterTextSplitter Statistics")
chunk_statistics(recursive_chunks)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

```python
docs = retriever.invoke("What is LangChain modularity?")

for i, doc in enumerate(docs, start=1):
    print(f"Chunk {i}")
    print(doc.page_content[:200])
    print("-"*60)
```

```python
def answer_question(question):
    docs = retriever.invoke(question)

    context = "\n\n".join(doc.page_content for doc in docs)

    prompt = f"""
Answer ONLY using the context below.

Context:
{context}

Question:
{question}
"""

    return ask_llm(prompt)
```

Multi-Step Processing with Traditional Chains vs LCEL

```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain, SequentialChain
```

```python
sentiment_prompt = PromptTemplate(
    input_variables=["review"],
    template="""
Analyze the sentiment of this product review.

Review:
{review}

Return only one word: Positive, Negative, or Neutral.
"""
)

summary_prompt = PromptTemplate(
    input_variables=["review"],
    template="""
Summarize this product review in one sentence.

Review:
{review}
"""
)

response_prompt = PromptTemplate(
    input_variables=["summary", "sentiment"],
    template="""
You are a customer support assistant.

Review Summary:
{summary}

Sentiment:
{sentiment}

Write a polite response to the customer.
"""
)
```

```python
sentiment_chain = LLMChain(
    llm=chat_llm,
    prompt=sentiment_prompt,
    output_key="sentiment"
)

summary_chain = LLMChain(
    llm=chat_llm,
    prompt=summary_prompt,
    output_key="summary"
)

response_chain = LLMChain(
    llm=chat_llm,
    prompt=response_prompt,
    output_key="response"
)
```

```python
traditional_chain = SequentialChain(
    chains=[sentiment_chain, summary_chain, response_chain],
    input_variables=["review"],
    output_variables=["sentiment", "summary", "response"],
    verbose=True
)
```

```python
review = """
The phone camera is excellent and battery life is amazing.
However, the charging speed could be better.
"""

result = traditional_chain.invoke({"review": review})

print("Sentiment:", result["sentiment"])
print("\nSummary:", result["summary"])
print("\nCustomer Response:\n", result["response"])
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
```

```python
lcel_sentiment = (
    sentiment_prompt
    | chat_llm
    | parser
)
```

```python
lcel_summary = (
    summary_prompt
    | chat_llm
    | parser
)
```

```python
lcel_response = (
    response_prompt
    | chat_llm
    | parser
)
```

```python
def lcel_review_processor(review):

    sentiment = lcel_sentiment.invoke({"review": review})

    summary = lcel_summary.invoke({"review": review})

    response = lcel_response.invoke({
        "summary": summary,
        "sentiment": sentiment
    })

    return {
        "sentiment": sentiment,
        "summary": summary,
        "response": response
    }
```

```python
review = """
The phone camera is excellent and battery life is amazing.
However, the charging speed could be better.
"""

result = lcel_review_processor(review)

print("Sentiment:", result["sentiment"])
print("\nSummary:", result["summary"])
print("\nCustomer Response:\n", result["response"])
```

```python
from langchain_core.tools import tool
from langgraph.prebuilt import create_react_agent
```

```python
@tool
def calculator(expression: str) -> str:
    """Perform mathematical calculations."""
    try:
        return str(eval(expression))
    except Exception:
        return "Invalid calculation."
```

```python
@tool
def formatter(text: str) -> str:
    """Convert text to uppercase."""
    return text.upper()
```

```python
# Reload Qwen and create chat_llm

from transformers import pipeline
from langchain_community.llms import HuggingFacePipeline
import langchain

# Compatibility patch
if not hasattr(langchain, "verbose"):
    langchain.verbose = False
if not hasattr(langchain, "debug"):
    langchain.debug = False

# Reload model (loads from cache if already downloaded)
llm = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=128,
    temperature=0.3,
    clean_up_tokenization_spaces=False,
)

# Convert to LangChain LLM
chat_llm = HuggingFacePipeline(pipeline=llm)

print("✅ llm and chat_llm are ready!")
```

```python
import re

def simple_agent(user_input):
    text = user_input.lower()

    # Tool 1: Formatter
    if "uppercase" in text or "capital" in text:
        match = re.search(r"'(.*?)'|\"(.*?)\"", user_input)
        if match:
            value = match.group(1) or match.group(2)
            return formatter.invoke({"text": value})
        return "Please provide text inside quotes."

    # Tool 2: Calculator
    if any(op in user_input for op in ["+", "-", "*", "/"]):
        expression = re.sub(r"[^0-9+\-*/(). ]", "", user_input)
        return calculator.invoke({"expression": expression})

    # Otherwise use the LLM
    result = llm(user_input, max_new_tokens=80)
    return result[0]["generated_text"]
```

```python
print(simple_agent("What is 25 * 8 + 12?"))
```

```python
print(simple_agent("Convert 'hello megan' to uppercase."))
```

```python
questions = [
    "What is 100 / 5 + 20?",
    "Convert 'artificial intelligence' to uppercase.",
    "What is 12 * 15?"
]

for q in questions:
    print("Question:", q)
    print("Answer:", simple_agent(q))
    print("-" * 60)
```
